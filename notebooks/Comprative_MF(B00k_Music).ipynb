```python
import pandas as pd

def parse_file_to_dataframe(file_path):
    # Initializing lists for each column
    ids, asins, titles, groups, salesranks, similars, customer_ids, ratings, votes, helpfuls = [], [], [], [], [], [], [], [], [], []

    # Temporary variables for current product
    current_id, current_asin, current_title, current_group, current_salesrank, current_similar = None, None, None, None, None, None

    # Flag to check if we are in the review section
    in_review_section = False

    with open(file_path, 'r') as file:
        for line in file:
            if line.startswith('Id'):
                # When a new product section starts, reset the flag and temporary variables
                in_review_section = False
                current_id, current_asin, current_title, current_group, current_salesrank, current_similar = None, None, None, None, None, None

            # Splitting line into key and value
            parts = line.split(':', 1)
            key = parts[0].strip() if len(parts) > 0 else ''
            value = parts[1].strip() if len(parts) > 1 else ''

            if key == 'Id':
                current_id = value
            elif key == 'ASIN':
                current_asin = value
            elif key == 'title':
                current_title = value
            elif key == 'group':
                current_group = value
            elif key == 'salesrank':
                current_salesrank = value
            elif key == 'similar':
                current_similar = value
            elif key == 'reviews':
                in_review_section = True  # Start of review section
                continue

            # Processing review lines
            if in_review_section and 'cutomer' in line:
                review_parts = line.split()
                if len(review_parts) >= 6:
                    # Appending data for each review
                    ids.append(current_id)
                    asins.append(current_asin)
                    titles.append(current_title)
                    groups.append(current_group)
                    salesranks.append(current_salesrank)
                    similars.append(current_similar)
                    customer_ids.append(review_parts[2])
                    ratings.append(review_parts[4])
                    votes.append(review_parts[6])
                    helpfuls.append(review_parts[8])

    # Creating the DataFrame
    data = {
        'ID': ids,
        'ASIN': asins,
        'Title': titles,
        'Group': groups,
        'Salesrank': salesranks,
        'Similar': similars,
        'Customerid': customer_ids,
        'Rating': ratings,
        'Votes': votes,
        'Helpful': helpfuls
    }
    df = pd.DataFrame(data)
    return df

# Usage example
file_path = '/content/drive/MyDrive/amazon-meta.txt'
df_reviews = parse_file_to_dataframe(file_path)
df_reviews.head()  # Displaying the first few rows of the DataFrame to verify the result
df_reviews.shape
```

# Revision extras: global-mean (μ) baseline
Run after `target_rating_matrix` is built. Uses the same 5-fold shuffle seed (42) and train fractions (20/40/60/80%) as the MF experiments.

```python
import numpy as np
from sklearn.utils import shuffle

def evaluate_mu_baseline(R_true, mu):
    mask = R_true > 0
    if not np.any(mask):
        return np.nan, np.nan
    err = R_true[mask] - mu
    return float(np.mean(np.abs(err))), float(np.sqrt(np.mean(err**2)))

def cross_validate_mu_baseline(target_rating_matrix, n_folds=5, random_state=42,
                               fractions=(0.2, 0.4, 0.6, 0.8)):
    R = target_rating_matrix.to_numpy() if hasattr(target_rating_matrix, 'to_numpy') else np.asarray(target_rating_matrix)
    pairs = [(u, i, R[u, i]) for u in range(R.shape[0]) for i in range(R.shape[1]) if R[u, i] > 0]
    pairs = shuffle(pairs, random_state=random_state)
    fold_size = len(pairs) // n_folds
    frac_mae = {f: [] for f in fractions}
    frac_rmse = {f: [] for f in fractions}
    full_mae, full_rmse = [], []

    for k in range(n_folds):
        te = pairs[k*fold_size:(k+1)*fold_size] if k < n_folds-1 else pairs[k*fold_size:]
        tr = [p for j,p in enumerate(pairs) if not (j >= k*fold_size and (k == n_folds-1 or j < (k+1)*fold_size))]
        # rebuild train/test matrices
        R_train = np.zeros_like(R)
        R_test = np.zeros_like(R)
        for u, i, r in tr:
            R_train[u, i] = r
        for u, i, r in te:
            R_test[u, i] = r
        mu_full = R_train[R_train > 0].mean() if np.any(R_train > 0) else 0.0
        mae, rmse = evaluate_mu_baseline(R_test, mu_full)
        full_mae.append(mae); full_rmse.append(rmse)
        print(f'Fold {k+1}: FULL - MAE={mae:.4f}, RMSE={rmse:.4f} (mu={mu_full:.4f})')
        idx = shuffle(list(range(len(tr))), random_state=random_state)
        for frac in fractions:
            cut = max(1, int(len(tr) * frac))
            partial = [tr[j] for j in idx[:cut]]
            R_part = np.zeros_like(R)
            for u, i, r in partial:
                R_part[u, i] = r
            mu = R_part[R_part > 0].mean() if np.any(R_part > 0) else 0.0
            mae_p, rmse_p = evaluate_mu_baseline(R_test, mu)
            frac_mae[frac].append(mae_p); frac_rmse[frac].append(rmse_p)
            print(f'Fold {k+1}: {int(frac*100)}% - MAE={mae_p:.4f}, RMSE={rmse_p:.4f} (mu={mu:.4f})')

    def summarize(name, maes, rmses):
        m = np.mean(maes); r = np.mean(rmses)
        mci = 1.96*np.std(maes, ddof=1)/np.sqrt(len(maes))
        rci = 1.96*np.std(rmses, ddof=1)/np.sqrt(len(rmses))
        print(f'=== {name} ===')
        print(f'MAE : {m:.4f} ± {mci:.4f}')
        print(f'RMSE: {r:.4f} ± {rci:.4f}')
        return m, mci, r, rci

    summarize('μ-baseline FULL', full_mae, full_rmse)
    results = {}
    for frac in fractions:
        results[frac] = summarize(f'μ-baseline {int(frac*100)}%', frac_mae[frac], frac_rmse[frac])
    return results

mu_baseline_results = cross_validate_mu_baseline(target_rating_matrix)
print(mu_baseline_results)
```

# Vanilla Matrix factorization

```python
import numpy as np
from sklearn.utils import shuffle

# ---------------------------------------------------
# Matrix Factorization and Transfer Learning Routines
# ---------------------------------------------------

def initialize_factors(n_users, n_items, n_factors):
    """
    Initialize latent factors with small random values.
    """
    P = np.random.normal(scale=0.1, size=(n_users, n_factors))
    Q = np.random.normal(scale=0.1, size=(n_items, n_factors))
    return P, Q

def train_matrix_factorization(R, P, Q, lambda_reg, learning_rate, epochs):
    """
    Train vanilla matrix factorization using SGD on ratings R.
    Only updates entries where R[u, i] > 0.
    """
    n_users, n_items = R.shape
    for _ in range(epochs):
        for u in range(n_users):
            for i in range(n_items):
                if R[u, i] > 0:
                    prediction = np.dot(P[u, :], Q[i, :])
                    error = R[u, i] - prediction

                    # Gradient update
                    P[u, :] += learning_rate * (error * Q[i, :] - lambda_reg * P[u, :])
                    Q[i, :] += learning_rate * (error * P[u, :] - lambda_reg * Q[i, :])
    return P, Q

def transfer_training(R_target, P_target, Q_target, P_source, Q_source,
                      lambda_reg, gamma, learning_rate, epochs,
                      common_users, common_items):
    """
    Fine-tune target domain latent factors with transfer regularization.
    """
    n_users, n_items = R_target.shape
    for _ in range(epochs):
        for u in range(n_users):
            for i in range(n_items):
                if R_target[u, i] > 0:
                    prediction = np.dot(P_target[u, :], Q_target[i, :])
                    error = R_target[u, i] - prediction

                    # Standard MF update
                    P_target[u, :] += learning_rate * (error * Q_target[i, :] - lambda_reg * P_target[u, :])
                    Q_target[i, :] += learning_rate * (error * P_target[u, :] - lambda_reg * Q_target[i, :])

                    # Transfer regularization for common users/items
                    if u in common_users:
                        P_target[u, :] -= learning_rate * gamma * (P_target[u, :] - P_source[u, :])
                    if i in common_items:
                        Q_target[i, :] -= learning_rate * gamma * (Q_target[i, :] - Q_source[i, :])
    return P_target, Q_target

def predict(P, Q):
    """
    Predict ratings as the dot product of the latent factor matrices.
    """
    return np.dot(P, Q.T)

def evaluate(R_true, R_pred):
    """
    Evaluate the predictions using MAE and RMSE on the observed ratings.
    """
    mask = (R_true > 0)
    errors = R_pred[mask] - R_true[mask]
    mae = np.mean(np.abs(errors))
    rmse = np.sqrt(np.mean(errors**2))
    return mae, rmse


# ------------------------------------------------------
# Example: Cross-Domain Transfer + K-Fold CV + Fractions
# ------------------------------------------------------

# Convert to NumPy if needed
source_rating_array = source_rating_matrix.to_numpy() if hasattr(source_rating_matrix, 'to_numpy') else source_rating_matrix
target_rating_array = target_rating_matrix.to_numpy() if hasattr(target_rating_matrix, 'to_numpy') else target_rating_matrix

n_users_source, n_items_source = source_rating_array.shape
n_users_target, n_items_target = target_rating_array.shape
assert n_users_source == n_users_target, "Source and target must have the same number of users."
n_users = n_users_source

n_factors = 10
lambda_reg = 0.01
learning_rate = 0.005
epochs_source = 20
epochs_target = 20
gamma = 0.1
k_folds = 5

# 1) Train on the source domain
P_source, Q_source = initialize_factors(n_users, n_items_source, n_factors)
P_source, Q_source = train_matrix_factorization(
    source_rating_array, P_source, Q_source,
    lambda_reg=lambda_reg, learning_rate=learning_rate,
    epochs=epochs_source
)

# Overlapping sets (example: all users, plus min of items)
common_users = list(range(n_users))
common_items = list(range(min(n_items_source, n_items_target)))

# Gather (u, i, r) from target domain
user_item_pairs = []
for u in range(n_users_target):
    for i in range(n_items_target):
        if target_rating_array[u, i] > 0:
            user_item_pairs.append((u, i, target_rating_array[u, i]))

user_item_pairs = shuffle(user_item_pairs, random_state=42)
fold_size = len(user_item_pairs) // k_folds
folds = []
for k in range(k_folds):
    start = k * fold_size
    end = start + fold_size
    folds.append(user_item_pairs[start:end])

mae_scores_full = []
rmse_scores_full = []

# We'll track partial fraction results in separate lists
fractions = [0.2, 0.4, 0.6, 0.8]
frac_mae_scores = {f: [] for f in fractions}
frac_rmse_scores = {f: [] for f in fractions}

# -------------------------------------------
# 2) 5-Fold Cross Validation
# -------------------------------------------
for k in range(k_folds):
    # Create train/test splits
    test_data = folds[k]
    train_data = []
    for j in range(k_folds):
        if j != k:
            train_data.extend(folds[j])

    # Build the FULL train and test rating matrices
    R_train_full = np.zeros((n_users_target, n_items_target))
    for (u, i, r) in train_data:
        R_train_full[u, i] = r

    R_test = np.zeros((n_users_target, n_items_target))
    for (u, i, r) in test_data:
        R_test[u, i] = r

    # ----------- A) FULL TRAINING -----------
    # Initialize target factors
    P_target_full, Q_target_full = initialize_factors(n_users_target, n_items_target, n_factors)

    # Transfer training on the FULL train set
    P_target_full, Q_target_full = transfer_training(
        R_train_full, P_target_full, Q_target_full,
        P_source, Q_source,
        lambda_reg=lambda_reg, gamma=gamma,
        learning_rate=learning_rate, epochs=epochs_target,
        common_users=common_users, common_items=common_items
    )

    # Evaluate on the test fold
    R_pred_full = predict(P_target_full, Q_target_full)
    mae_full, rmse_full = evaluate(R_test, R_pred_full)
    mae_scores_full.append(mae_full)
    rmse_scores_full.append(rmse_full)

    print(f"\nFold {k+1}: FULL - MAE={mae_full:.4f}, RMSE={rmse_full:.4f}")

    # ----------- B) PARTIAL FRACTIONS -----------
    # Shuffle 'train_data' so we can sub-sample randomly
    rng = np.random.default_rng(42)
    indices = np.arange(len(train_data))
    rng.shuffle(indices)

    for frac in fractions:
        cutoff = int(len(train_data) * frac)
        partial_indices = indices[:cutoff]
        partial_data = [train_data[idx] for idx in partial_indices]

        # Build a partial R_train matrix
        R_train_part = np.zeros((n_users_target, n_items_target))
        for (u, i, r) in partial_data:
            R_train_part[u, i] = r

        # Initialize target factors again
        P_target_part, Q_target_part = initialize_factors(n_users_target, n_items_target, n_factors)

        # Transfer training on this partial set
        P_target_part, Q_target_part = transfer_training(
            R_train_part, P_target_part, Q_target_part,
            P_source, Q_source,
            lambda_reg=lambda_reg, gamma=gamma,
            learning_rate=learning_rate, epochs=epochs_target,
            common_users=common_users, common_items=common_items
        )

        # Evaluate
        R_pred_part = predict(P_target_part, Q_target_part)
        mae_part, rmse_part = evaluate(R_test, R_pred_part)

        frac_mae_scores[frac].append(mae_part)
        frac_rmse_scores[frac].append(rmse_part)

        print(f"Fold {k+1}: {int(frac*100)}% - MAE={mae_part:.4f}, RMSE={rmse_part:.4f}")


# --------------------------------------------------
# 3) Aggregate Metrics and 95% Confidence Intervals
# --------------------------------------------------
def mean_std_confidence(values, z=1.96):
    m = np.mean(values)
    s = np.std(values, ddof=1)
    n = float(len(values))
    ci = z * (s / np.sqrt(n))
    return m, s, ci

mae_mean_full, _, mae_ci_full = mean_std_confidence(mae_scores_full)
rmse_mean_full, _, rmse_ci_full = mean_std_confidence(rmse_scores_full)

print("\n=== FINAL 5-Fold (FULL training) ===")
print(f"MAE : {mae_mean_full:.4f} ± {mae_ci_full:.4f}")
print(f"RMSE: {rmse_mean_full:.4f} ± {rmse_ci_full:.4f}")

# Summaries for each fraction
for frac in fractions:
    mae_vals = frac_mae_scores[frac]
    rmse_vals = frac_rmse_scores[frac]
    mae_m, _, mae_ci = mean_std_confidence(mae_vals)
    rmse_m, _, rmse_ci = mean_std_confidence(rmse_vals)

    print(f"\n=== FINAL 5-Fold ({int(frac*100)}% partial) ===")
    print(f"MAE : {mae_m:.4f} ± {mae_ci:.4f}")
    print(f"RMSE: {rmse_m:.4f} ± {rmse_ci:.4f}")
```

```python
import numpy as np
from sklearn.utils import shuffle

# ============================
# Helpers: Bias + MF
# ============================

def compute_global_mean(R):
    """Compute the global average rating (non-zero entries)."""
    nonzero = (R > 0)
    return R[nonzero].mean() if np.any(nonzero) else 0.0

def initialize_biases(n_users, n_items):
    """
    Initialize user and item biases to zero (or small random).
    """
    w_user = np.zeros(n_users, dtype=np.float32)
    w_item = np.zeros(n_items, dtype=np.float32)
    return w_user, w_item

def initialize_factors(n_users, n_items, n_factors):
    """
    Initialize latent factors with small random values.
    """
    P = np.random.normal(scale=0.1, size=(n_users, n_factors))
    Q = np.random.normal(scale=0.1, size=(n_items, n_factors))
    return P, Q

# -----------------------------------
# Train MF w/ Biases on Source Domain
# -----------------------------------
def train_mf_bias(
    R_source,
    n_factors=10,
    lambda_reg=0.01,
    learning_rate=0.005,
    epochs=20
):
    """
    Train matrix factorization with user/item biases on the source domain.

    Minimizes:
      Σ (r_ui - (b + w_u + w_i + p_u^T q_i))^2 + λ( ... )
    """

    n_users, n_items = R_source.shape
    # Global average
    b = compute_global_mean(R_source)
    # Initialize biases
    w_user, w_item = initialize_biases(n_users, n_items)
    # Initialize latent factors
    P, Q = initialize_factors(n_users, n_items, n_factors)

    for _ in range(epochs):
        for u in range(n_users):
            for i in range(n_items):
                r_ui = R_source[u, i]
                if r_ui > 0:
                    # Prediction
                    pred = b + w_user[u] + w_item[i] + np.dot(P[u, :], Q[i, :])
                    err = r_ui - pred

                    # Update biases
                    w_user[u] += learning_rate * (err - lambda_reg * w_user[u])
                    w_item[i] += learning_rate * (err - lambda_reg * w_item[i])

                    # Update latent factors
                    Pu_old = P[u, :].copy()  # for simultaneous update
                    P[u, :] += learning_rate * (err * Q[i, :] - lambda_reg * P[u, :])
                    Q[i, :] += learning_rate * (err * Pu_old   - lambda_reg * Q[i, :])

        # Optional: If you want to treat b as a parameter, you'd do:
        # b += learning_rate * (err - lambda_reg * b)
        # but typically we keep b fixed as the average rating.

    return b, w_user, w_item, P, Q

# -----------------------------------
# Transfer Training (Target Domain)
# -----------------------------------
def transfer_training_bias(
    R_target,
    b_source, w_user_source, w_item_source, P_source, Q_source,
    # We'll also have our own target biases/factors:
    b_target, w_user_target, w_item_target, P_target, Q_target,
    lambda_reg=0.01, gamma=0.1, learning_rate=0.005, epochs=20,
    common_users=None, common_items=None,
    do_bias_transfer=True
):
    """
    Fine-tune target domain (b_target, w_user_target, w_item_target, P_target, Q_target)
    with transfer regularization from the source domain.

    Minimizes for each known rating (u,i):
        (r_ui - (b_t + w_user_t[u] + w_item_t[i] + P_t[u]^T Q_t[i]))^2
        + λ(...) + γ(...)   # transfer penalty

    If do_bias_transfer=True, we also add a penalty for user/item bias differences:
        γ (Σ_{u in commonU} ||P_t[u] - P_s[u]||^2
           + Σ_{i in commonI} ||Q_t[i] - Q_s[i]||^2
           + Σ_{u in commonU} (w_user_t[u] - w_user_s[u])^2
           + Σ_{i in commonI} (w_item_t[i] - w_item_s[i])^2)
    """
    n_users, n_items = R_target.shape
    common_users = set(common_users) if common_users else set()
    common_items = set(common_items) if common_items else set()

    for _ in range(epochs):
        for u in range(n_users):
            for i in range(n_items):
                r_ui = R_target[u, i]
                if r_ui > 0:
                    # Prediction
                    pred = b_target + w_user_target[u] + w_item_target[i] + np.dot(P_target[u], Q_target[i])
                    err = r_ui - pred

                    # Update biases
                    w_user_target[u] += learning_rate * (err - lambda_reg * w_user_target[u])
                    w_item_target[i] += learning_rate * (err - lambda_reg * w_item_target[i])

                    # Update latent factors (simultaneous update trick)
                    Pu_old = P_target[u].copy()
                    P_target[u] += learning_rate * (err * Q_target[i] - lambda_reg * P_target[u])
                    Q_target[i] += learning_rate * (err * Pu_old        - lambda_reg * Q_target[i])

                    # Transfer regularization for common users/items
                    if u in common_users:
                        # Pull P_target[u] toward P_source[u]
                        diff_p = (P_target[u] - P_source[u])
                        P_target[u] -= learning_rate * gamma * diff_p

                        # If also transferring biases:
                        if do_bias_transfer:
                            diff_bu = (w_user_target[u] - w_user_source[u])
                            w_user_target[u] -= learning_rate * gamma * diff_bu

                    if i in common_items:
                        # Pull Q_target[i] toward Q_source[i]
                        diff_q = (Q_target[i] - Q_source[i])
                        Q_target[i] -= learning_rate * gamma * diff_q

                        if do_bias_transfer:
                            diff_bi = (w_item_target[i] - w_item_source[i])
                            w_item_target[i] -= learning_rate * gamma * diff_bi

        # Optionally treat b_target as updatable:
        # b_target += learning_rate * (...)
        # Typically we keep it fixed to the average of target ratings.

    return (b_target, w_user_target, w_item_target, P_target, Q_target)

# ===============================
# Predict & Evaluate
# ===============================
def predict(b, w_user, w_item, P, Q):
    """Full prediction: b + w_user[u] + w_item[i] + P[u] dot Q[i]."""
    # shape: (n_users, n_items)
    pred = b + w_user[:, np.newaxis] + w_item[np.newaxis, :] + P @ Q.T
    return pred

def evaluate(R_true, R_pred):
    """
    Evaluate predictions where R_true > 0, returning (MAE, RMSE).
    """
    mask = (R_true > 0)
    errors = R_pred[mask] - R_true[mask]
    mae = np.mean(np.abs(errors))
    rmse = np.sqrt(np.mean(errors**2))
    return mae, rmse


# ===============================
# Example: Cross-Validation
# ===============================
def cross_validate_transfer_mf_bias(
    source_rating_matrix,
    target_rating_matrix,
    n_factors=10,
    lambda_reg=0.01,
    gamma=0.1,
    learning_rate=0.005,
    epochs_source=20,
    epochs_target=20,
    k_folds=5
):
    # Convert to NumPy if needed
    if hasattr(source_rating_matrix, "to_numpy"):
        R_source = source_rating_matrix.to_numpy()
    else:
        R_source = source_rating_matrix

    if hasattr(target_rating_matrix, "to_numpy"):
        R_target = target_rating_matrix.to_numpy()
    else:
        R_target = target_rating_matrix

    n_users_source, n_items_source = R_source.shape
    n_users_target, n_items_target = R_target.shape
    assert n_users_source == n_users_target, "User sets must match!"
    n_users = n_users_source

    # ---------------------
    # 1) Train on Source
    # ---------------------
    b_source, w_user_source, w_item_source, P_source, Q_source = train_mf_bias(
        R_source,
        n_factors=n_factors,
        lambda_reg=lambda_reg,
        learning_rate=learning_rate,
        epochs=epochs_source
    )

    # We'll define the common user/item sets
    common_users = list(range(n_users))
    common_items = list(range(min(n_items_source, n_items_target)))

    # Build (u,i,r) triples for target domain
    triples = []
    for u in range(n_users_target):
        for i in range(n_items_target):
            if R_target[u, i] > 0:
                triples.append((u, i, R_target[u, i]))

    # Shuffle and create k folds
    triples = shuffle(triples, random_state=42)
    fold_size = len(triples) // k_folds
    folds = []
    for k in range(k_folds):
        start = k * fold_size
        end = start + fold_size
        folds.append(triples[start:end])

    mae_scores, rmse_scores = [], []

    # ---------------------
    # 2) K-Fold CV
    # ---------------------
    for k in range(k_folds):
        # Split train/test
        test_data = folds[k]
        train_data = []
        for j in range(k_folds):
            if j != k:
                train_data.extend(folds[j])

        R_train = np.zeros_like(R_target)
        R_test  = np.zeros_like(R_target)
        for (u, i, r) in train_data:
            R_train[u, i] = r
        for (u, i, r) in test_data:
            R_test[u, i] = r

        # Initialize target biases/factors from scratch
        b_target = compute_global_mean(R_train)  # or 0.0
        w_user_target, w_item_target = initialize_biases(n_users_target, n_items_target)
        P_target, Q_target = initialize_factors(n_users_target, n_items_target, n_factors)

        # Transfer training
        b_target, w_user_target, w_item_target, P_target, Q_target = transfer_training_bias(
            R_train,
            b_source, w_user_source, w_item_source, P_source, Q_source,
            b_target, w_user_target, w_item_target, P_target, Q_target,
            lambda_reg=lambda_reg, gamma=gamma,
            learning_rate=learning_rate, epochs=epochs_target,
            common_users=common_users, common_items=common_items,
            do_bias_transfer=True  # set to False if you don't want to transfer biases
        )

        # Evaluate on test fold
        R_pred = predict(b_target, w_user_target, w_item_target, P_target, Q_target)
        mae, rmse = evaluate(R_test, R_pred)
        mae_scores.append(mae)
        rmse_scores.append(rmse)

        print(f"Fold {k+1}: MAE={mae:.4f}, RMSE={rmse:.4f}")

    # ---------------------
    # 3) 95% Confidence Intervals
    # ---------------------
    mae_mean, rmse_mean = np.mean(mae_scores), np.mean(rmse_scores)
    mae_std, rmse_std = np.std(mae_scores, ddof=1), np.std(rmse_scores, ddof=1)
    z_value = 1.96
    n_folds = float(k_folds)

    mae_ci   = z_value * (mae_std  / np.sqrt(n_folds))
    rmse_ci  = z_value * (rmse_std / np.sqrt(n_folds))

    print("\n=== Final 5-Fold Results (Target Domain) ===")
    print(f"MAE:  {mae_mean:.4f} ± {mae_ci:.4f}  (95% CI)")
    print(f"RMSE: {rmse_mean:.4f} ± {rmse_ci:.4f}  (95% CI)")


# ===============================
# Usage Example
# ===============================
# Suppose you have:
#   source_rating_matrix, target_rating_matrix
# Just call:
# If your source_rating_matrix and target_rating_matrix are already defined
# (like in your display code) as NumPy arrays or DataFrames, you just do:

cross_validate_transfer_mf_bias(
    source_rating_matrix=source_rating_matrix,
    target_rating_matrix=target_rating_matrix,
    n_factors=10,
    lambda_reg=0.01,
    gamma=0.1,
    learning_rate=0.005,
    epochs_source=20,
    epochs_target=20,
    k_folds=5
)

#
# This will:
#   1) Train MF-with-bias on the full source.
#   2) 5-fold cross-validate in the target domain with transfer from the source factors + biases.
#   3) Print out fold-level performance + final average ± 95% CI.
```

```python
import numpy as np
from sklearn.utils import shuffle

# ============================
# Helpers: Bias + MF
# ============================

def compute_global_mean(R):
    """Compute the global average rating (non-zero entries)."""
    nonzero = (R > 0)
    return R[nonzero].mean() if np.any(nonzero) else 0.0

def initialize_biases(n_users, n_items):
    """Initialize user and item biases (0.0 or small random)."""
    w_user = np.zeros(n_users, dtype=np.float32)
    w_item = np.zeros(n_items, dtype=np.float32)
    return w_user, w_item

def initialize_factors(n_users, n_items, n_factors):
    """Initialize latent factors with small random values."""
    P = np.random.normal(scale=0.1, size=(n_users, n_factors))
    Q = np.random.normal(scale=0.1, size=(n_items, n_factors))
    return P, Q

# -----------------------------------
# Train MF w/ Biases on Source Domain
# -----------------------------------
def train_mf_bias(
    R_source,
    n_factors=10,
    lambda_reg=0.01,
    learning_rate=0.005,
    epochs=20
):
    """
    Train matrix factorization with user/item biases on the source domain.

    Minimizes:
      Σ (r_ui - (b + w_u + w_i + p_u^T q_i))^2 + λ( ... )
    """
    n_users, n_items = R_source.shape
    # Global average
    b = compute_global_mean(R_source)
    # Initialize biases
    w_user, w_item = initialize_biases(n_users, n_items)
    # Initialize latent factors
    P, Q = initialize_factors(n_users, n_items, n_factors)

    for _ in range(epochs):
        for u in range(n_users):
            for i in range(n_items):
                r_ui = R_source[u, i]
                if r_ui > 0:
                    # Prediction
                    pred = b + w_user[u] + w_item[i] + np.dot(P[u, :], Q[i, :])
                    err = r_ui - pred

                    # Update biases
                    w_user[u] += learning_rate * (err - lambda_reg * w_user[u])
                    w_item[i] += learning_rate * (err - lambda_reg * w_item[i])

                    # Update latent factors
                    Pu_old = P[u, :].copy()  # for simultaneous update
                    P[u, :] += learning_rate * (err * Q[i, :] - lambda_reg * P[u, :])
                    Q[i, :] += learning_rate * (err * Pu_old   - lambda_reg * Q[i, :])

    return b, w_user, w_item, P, Q

# -----------------------------------
# Transfer Training (Target Domain)
# -----------------------------------
def transfer_training_bias(
    R_target,
    b_source, w_user_source, w_item_source, P_source, Q_source,
    # We'll also have our own target biases/factors:
    b_target, w_user_target, w_item_target, P_target, Q_target,
    lambda_reg=0.01, gamma=0.1, learning_rate=0.005, epochs=20,
    common_users=None, common_items=None,
    do_bias_transfer=True
):
    """
    Fine-tune target domain (b_target, w_user_target, w_item_target, P_target, Q_target)
    with transfer regularization from the source domain.

    Minimizes for each known rating (u,i):
        (r_ui - (b_t + w_user_t[u] + w_item_t[i] + P_t[u]^T Q_t[i]))^2
        + λ(...) + γ(...)   # transfer penalty

    If do_bias_transfer=True, we also add a penalty for user/item bias differences.
    """
    n_users, n_items = R_target.shape
    common_users = set(common_users) if common_users else set()
    common_items = set(common_items) if common_items else set()

    for _ in range(epochs):
        for u in range(n_users):
            for i in range(n_items):
                r_ui = R_target[u, i]
                if r_ui > 0:
                    # Prediction
                    pred = b_target + w_user_target[u] + w_item_target[i] + np.dot(P_target[u], Q_target[i])
                    err = r_ui - pred

                    # Update biases
                    w_user_target[u] += learning_rate * (err - lambda_reg * w_user_target[u])
                    w_item_target[i] += learning_rate * (err - lambda_reg * w_item_target[i])

                    # Update latent factors (simultaneous update)
                    Pu_old = P_target[u].copy()
                    P_target[u] += learning_rate * (err * Q_target[i] - lambda_reg * P_target[u])
                    Q_target[i] += learning_rate * (err * Pu_old        - lambda_reg * Q_target[i])

                    # Transfer regularization for common users/items
                    if u in common_users:
                        # Pull P_target[u] toward P_source[u]
                        diff_p = (P_target[u] - P_source[u])
                        P_target[u] -= learning_rate * gamma * diff_p

                        if do_bias_transfer:
                            diff_bu = (w_user_target[u] - w_user_source[u])
                            w_user_target[u] -= learning_rate * gamma * diff_bu

                    if i in common_items:
                        diff_q = (Q_target[i] - Q_source[i])
                        Q_target[i] -= learning_rate * gamma * diff_q

                        if do_bias_transfer:
                            diff_bi = (w_item_target[i] - w_item_source[i])
                            w_item_target[i] -= learning_rate * gamma * diff_bi

    return (b_target, w_user_target, w_item_target, P_target, Q_target)

# ===============================
# Predict & Evaluate
# ===============================
def predict(b, w_user, w_item, P, Q):
    """Full prediction: b + w_user[u] + w_item[i] + P[u] dot Q[i]."""
    return b + w_user[:, np.newaxis] + w_item[np.newaxis, :] + P @ Q.T

def evaluate(R_true, R_pred):
    """Evaluate predictions where R_true > 0, returning (MAE, RMSE)."""
    mask = (R_true > 0)
    errors = R_pred[mask] - R_true[mask]
    mae = np.mean(np.abs(errors))
    rmse = np.sqrt(np.mean(errors**2))
    return mae, rmse


# ===============================
# Example: K-Fold CV + Partial Fractions
# ===============================
def cross_validate_transfer_mf_bias(
    source_rating_matrix,
    target_rating_matrix,
    n_factors=10,
    lambda_reg=0.01,
    gamma=0.1,
    learning_rate=0.005,
    epochs_source=20,
    epochs_target=20,
    k_folds=5
):
    """
    1) Train MF-with-bias on the full source domain.
    2) K-fold cross-validation on the target domain with transfer from source.
    3) Evaluate performance on the entire training set (as in the original code),
       then also sub-sample 20%, 40%, 60%, 80% of the training data to see how
       performance changes with less target-domain data.
    """
    # Convert to NumPy if needed
    if hasattr(source_rating_matrix, "to_numpy"):
        R_source = source_rating_matrix.to_numpy()
    else:
        R_source = source_rating_matrix

    if hasattr(target_rating_matrix, "to_numpy"):
        R_target = target_rating_matrix.to_numpy()
    else:
        R_target = target_rating_matrix

    n_users_source, n_items_source = R_source.shape
    n_users_target, n_items_target = R_target.shape
    # We assume same user dimension if you are transferring user factors
    assert n_users_source == n_users_target, "User sets must match!"
    n_users = n_users_source

    # ---------------------
    # 1) Train on Source
    # ---------------------
    b_source, w_user_source, w_item_source, P_source, Q_source = train_mf_bias(
        R_source,
        n_factors=n_factors,
        lambda_reg=lambda_reg,
        learning_rate=learning_rate,
        epochs=epochs_source
    )

    # We'll define the common user/item sets
    common_users = list(range(n_users))
    common_items = list(range(min(n_items_source, n_items_target)))

    # Build (u,i,r) triples for target domain
    triples = []
    for u in range(n_users_target):
        for i in range(n_items_target):
            if R_target[u, i] > 0:
                triples.append((u, i, R_target[u, i]))

    # Shuffle and create k folds
    triples = shuffle(triples, random_state=42)
    fold_size = len(triples) // k_folds
    folds = []
    for k in range(k_folds):
        start = k * fold_size
        end = start + fold_size
        folds.append(triples[start:end])

    # We'll store the standard full-training performance
    mae_scores_full, rmse_scores_full = [], []

    # For partial fraction experiments (20%, 40%, 60%, 80%):
    partial_fractions = [0.2, 0.4, 0.6, 0.8]
    # We'll store fold-level performance for each fraction
    fraction_mae = {frac: [] for frac in partial_fractions}
    fraction_rmse = {frac: [] for frac in partial_fractions}

    # ---------------------
    # 2) K-Fold CV
    # ---------------------
    for k in range(k_folds):
        # Split train/test
        test_data = folds[k]
        train_data = []
        for j in range(k_folds):
            if j != k:
                train_data.extend(folds[j])

        # Build the FULL training matrices
        R_train_full = np.zeros_like(R_target)
        R_test       = np.zeros_like(R_target)
        for (u, i, r) in train_data:
            R_train_full[u, i] = r
        for (u, i, r) in test_data:
            R_test[u, i] = r

        # -- (A) Standard Training on FULL train_data --
        b_target = compute_global_mean(R_train_full)
        w_user_target, w_item_target = initialize_biases(n_users_target, n_items_target)
        P_target, Q_target = initialize_factors(n_users_target, n_items_target, n_factors)

        (b_target, w_user_target, w_item_target, P_target, Q_target) = transfer_training_bias(
            R_train_full,
            b_source, w_user_source, w_item_source, P_source, Q_source,
            b_target, w_user_target, w_item_target, P_target, Q_target,
            lambda_reg=lambda_reg, gamma=gamma,
            learning_rate=learning_rate, epochs=epochs_target,
            common_users=common_users, common_items=common_items,
            do_bias_transfer=True
        )

        # Evaluate on test fold
        R_pred_full = predict(b_target, w_user_target, w_item_target, P_target, Q_target)
        mae_full, rmse_full = evaluate(R_test, R_pred_full)
        mae_scores_full.append(mae_full)
        rmse_scores_full.append(rmse_full)

        print(f"\nFold {k+1} (FULL): MAE={mae_full:.4f}, RMSE={rmse_full:.4f}")

        # -- (B) Partial-Fraction Training --
        # We'll shuffle train_data, then sub-sample 20%, 40%, 60%, 80%.
        train_data_shuffled = train_data.copy()
        train_data_shuffled = shuffle(train_data_shuffled, random_state=42)  # ensure consistent ordering

        for frac in partial_fractions:
            cutoff = int(len(train_data_shuffled) * frac)
            partial_data = train_data_shuffled[:cutoff]

            # Build a partial training matrix
            R_train_partial = np.zeros_like(R_target)
            for (u, i, r) in partial_data:
                R_train_partial[u, i] = r

            # Initialize from scratch
            b_target_part = compute_global_mean(R_train_partial)
            w_user_target_part, w_item_target_part = initialize_biases(n_users_target, n_items_target)
            P_target_part, Q_target_part = initialize_factors(n_users_target, n_items_target, n_factors)

            # Transfer training on partial set
            (b_target_part, w_user_target_part, w_item_target_part,
             P_target_part, Q_target_part) = transfer_training_bias(
                R_train_partial,
                b_source, w_user_source, w_item_source, P_source, Q_source,
                b_target_part, w_user_target_part, w_item_target_part, P_target_part, Q_target_part,
                lambda_reg=lambda_reg, gamma=gamma,
                learning_rate=learning_rate, epochs=epochs_target,
                common_users=common_users, common_items=common_items,
                do_bias_transfer=True
            )

            # Evaluate on the same test fold
            R_pred_part = predict(b_target_part, w_user_target_part, w_item_target_part, P_target_part, Q_target_part)
            mae_part, rmse_part = evaluate(R_test, R_pred_part)
            fraction_mae[frac].append(mae_part)
            fraction_rmse[frac].append(rmse_part)

            print(f"Fold {k+1} ({int(frac*100)}%): MAE={mae_part:.4f}, RMSE={rmse_part:.4f}")

    # ---------------------
    # 3) Summaries + 95% CI
    # ---------------------
    def mean_and_ci(values):
        """Compute mean and 95% CI with normal approximation (z=1.96)."""
        arr = np.array(values)
        mean_val = arr.mean()
        std_val = arr.std(ddof=1)
        n = float(len(arr))
        z = 1.96
        margin = z * (std_val / np.sqrt(n))
        return mean_val, margin

    # Full data results
    mae_mean_full, mae_ci_full = mean_and_ci(mae_scores_full)
    rmse_mean_full, rmse_ci_full = mean_and_ci(rmse_scores_full)

    print("\n=== Final 5-Fold Results (FULL training) ===")
    print(f"MAE:  {mae_mean_full:.4f} ± {mae_ci_full:.4f}")
    print(f"RMSE: {rmse_mean_full:.4f} ± {rmse_ci_full:.4f}")

    # Fraction results
    for frac in partial_fractions:
        mae_mean_frac, mae_ci_frac = mean_and_ci(fraction_mae[frac])
        rmse_mean_frac, rmse_ci_frac = mean_and_ci(fraction_rmse[frac])
        print(f"\n=== Final 5-Fold Results (Training at {int(frac*100)}%) ===")
        print(f"MAE:  {mae_mean_frac:.4f} ± {mae_ci_frac:.4f}")
        print(f"RMSE: {rmse_mean_frac:.4f} ± {rmse_ci_frac:.4f}")


# ===============================
# Usage Example
# ===============================
# Suppose you have:
#   source_rating_matrix, target_rating_matrix
# as NumPy arrays or DataFrames.

# Just call:
cross_validate_transfer_mf_bias(
    source_rating_matrix=source_rating_matrix,
    target_rating_matrix=target_rating_matrix,
    n_factors=10,
    lambda_reg=0.01,
    gamma=0.1,
    learning_rate=0.005,
    epochs_source=20,
    epochs_target=20,
    k_folds=5
)
```

```python
import numpy as np
from sklearn.utils import shuffle

# =============================================================================
# 1) Train NMF on Source (Lee & Seung) with L2 regularization
# =============================================================================
def train_nmf_regularized(
    R_source,
    n_factors=10,
    alpha=0.01,      # L2 reg
    max_iter=20,
    tol=1e-4,
    random_state=42
):
    """
    Standard Lee & Seung multiplicative updates with L2 regularization:
      minimize 0.5 * ||R_source - W_s H_s||^2
             + 0.5 * alpha (||W_s||^2 + ||H_s||^2).
    Returns (W_source, H_source).
    """
    if hasattr(R_source, "to_numpy"):
        R_source = R_source.to_numpy()  # ensure NumPy

    np.random.seed(random_state)
    n_users, n_items = R_source.shape

    # Initialize nonnegative
    W_source = np.random.rand(n_users, n_factors)
    H_source = np.random.rand(n_factors, n_items)

    def frob_norm_sq(X):
        return np.sum(X**2)

    prev_loss = None
    for _ in range(max_iter):
        # Update H
        numerator_H   = W_source.T @ R_source
        denominator_H = (W_source.T @ W_source) @ H_source + alpha * H_source
        denominator_H[denominator_H == 0] = 1e-10
        H_source *= (numerator_H / denominator_H)

        # Update W
        numerator_W   = R_source @ H_source.T
        denominator_W = W_source @ (H_source @ H_source.T) + alpha * W_source
        denominator_W[denominator_W == 0] = 1e-10
        W_source *= (numerator_W / denominator_W)

        # Optional converge check
        WH = W_source @ H_source
        recon = 0.5 * frob_norm_sq(R_source - WH)
        reg   = 0.5 * alpha * (frob_norm_sq(W_source) + frob_norm_sq(H_source))
        loss  = recon + reg
        if prev_loss is not None and abs(prev_loss - loss) < tol:
            break
        prev_loss = loss

    return W_source, H_source


# =============================================================================
# 2) Transfer training on Target with a gamma penalty
#    pulling (W_target,H_target) toward (W_source,H_source).
#    Still uses Lee & Seung multiplicative updates (plus modifications).
# =============================================================================
def train_nmf_transfer(
    R_target,
    W_source, H_source,
    common_users,   # indices of shared users
    common_items,   # indices of shared items
    n_factors=10,
    alpha=0.01,     # L2 reg on target
    gamma=0.1,      # transfer penalty
    max_iter=20,
    tol=1e-4,
    random_state=123
):
    """
    Minimizes:
      0.5 * ||R_target - W_t H_t||^2
    + 0.5 * alpha (||W_t||^2 + ||H_t||^2)
    + 0.5 * gamma [
         Σ_{u in commonU}||W_t[u]-W_s[u]||^2 +
         Σ_{i in commonI}||H_t[:,i]-H_s[:,i]||^2 ].
    """
    if hasattr(R_target, "to_numpy"):
        R_target = R_target.to_numpy()

    np.random.seed(random_state)
    m, n = R_target.shape

    # Initialize target factors
    W_target = np.random.rand(m, n_factors)
    H_target = np.random.rand(n_factors, n)

    # Create binary masks for common rows/cols
    user_mask = np.zeros(m)
    user_mask[common_users] = 1.0
    item_mask = np.zeros(n)
    item_mask[common_items] = 1.0

    def frob_norm_sq(X):
        return np.sum(X**2)

    prev_loss = None
    for _ in range(max_iter):
        # Update H_target
        numerator_H   = W_target.T @ R_target
        denominator_H = (W_target.T @ W_target) @ H_target + alpha * H_target

        # Add transfer for columns in common
        numerator_H += gamma * (H_source * item_mask[np.newaxis, :])
        denominator_H += gamma * (H_target * item_mask[np.newaxis, :])

        denominator_H[denominator_H == 0] = 1e-10
        H_target *= (numerator_H / denominator_H)

        # Update W_target
        numerator_W   = R_target @ H_target.T
        denominator_W = W_target @ (H_target @ H_target.T) + alpha * W_target

        # Add transfer for rows in common
        numerator_W += gamma * (W_source * user_mask[:, np.newaxis])
        denominator_W += gamma * (W_target * user_mask[:, np.newaxis])

        denominator_W[denominator_W == 0] = 1e-10
        W_target *= (numerator_W / denominator_W)

        # Optional: compute total loss
        WtHt = W_target @ H_target
        recon = 0.5 * frob_norm_sq(R_target - WtHt)
        reg   = 0.5 * alpha * (frob_norm_sq(W_target) + frob_norm_sq(H_target))
        W_diff = (W_target - W_source)[common_users]  # only compare common user rows
        H_diff = (H_target - H_source)[:, common_items]  # only compare common item cols
        trans  = 0.5 * gamma * (frob_norm_sq(W_diff) + frob_norm_sq(H_diff))
        loss   = recon + reg + trans

        if prev_loss is not None and abs(prev_loss - loss) < tol:
            break
        prev_loss = loss

    return W_target, H_target


# =============================================================================
# 3) Prediction & Evaluation
# =============================================================================
def predict_nmf(W, H):
    """Compute R_pred = W @ H."""
    return W @ H

def evaluate(R_true, R_pred):
    """
    Evaluate only where R_true > 0, returning (MAE, RMSE).
    """
    mask = (R_true > 0)
    errors = R_pred[mask] - R_true[mask]
    mae = np.mean(np.abs(errors))
    rmse = np.sqrt(np.mean(errors**2))
    return mae, rmse


# =============================================================================
# 4) Cross-validation with shape fix for item mismatch
# =============================================================================
def cross_validate_transfer_nmf(
    source_rating_matrix,
    target_rating_matrix,
    n_factors=10,
    alpha=0.01,
    gamma=0.1,
    max_iter=20,
    k_folds=5,
    random_state=42
):
    """
    1) Train (W_source,H_source) on the FULL source domain.
    2) Possibly slice columns if #items differs between source & target
       so shapes match for 'common_items' dimension.
    3) k-fold cross-validation on the target domain,
       training with a penalty pulling (W_target,H_target)
       to (W_source,H_source).
    """
    # Convert to arrays
    if hasattr(source_rating_matrix, "to_numpy"):
        R_s = source_rating_matrix.to_numpy()
    else:
        R_s = source_rating_matrix

    if hasattr(target_rating_matrix, "to_numpy"):
        R_t = target_rating_matrix.to_numpy()
    else:
        R_t = target_rating_matrix

    # 1) Train on Source
    print("=== Train NMF on Source (Lee & Seung) ===")
    W_source, H_source = train_nmf_regularized(
        R_source=R_s,
        n_factors=n_factors,
        alpha=alpha,
        max_iter=max_iter,
        random_state=random_state
    )
    # Evaluate on source
    R_s_pred = predict_nmf(W_source, H_source)
    mae_s, rmse_s = evaluate(R_s, R_s_pred)
    print(f"Source Fit => MAE={mae_s:.4f}, RMSE={rmse_s:.4f}\n")

    # 2) Fix item-dimension mismatch if needed
    m_s, n_s = R_s.shape
    m_t, n_t = R_t.shape
    assert m_s == m_t, "User/row count mismatch between source & target!"

    if n_s > n_t:
        # Slice H_source to the first n_t columns
        # so 'common_items' can match the target domain
        H_source = H_source[:, :n_t]
        common_items = list(range(n_t))
        # => R_t keeps shape (m_t, n_t)
        # The leftover columns in H_source are discarded
        # because they have no counterpart in target.
    else:
        # n_t >= n_s => slice target matrix to shape (m_t, n_s),
        # so it lines up with the source domain's item dimension
        R_t = R_t[:, :n_s]
        common_items = list(range(n_s))
        # => H_source remains shape (n_factors, n_s)

    # Now, R_t has shape (m_t, <= original n_t),
    # and H_source has shape (n_factors, <= original n_s),
    # ensuring they have the same # of item columns.

    # 3) Build (u,i,r) in the possibly-sliced target
    m_t, n_t_sliced = R_t.shape
    triples = []
    for u in range(m_t):
        for i in range(n_t_sliced):
            if R_t[u, i] > 0:
                triples.append((u, i, R_t[u, i]))

    # Shuffle & fold
    triples = shuffle(triples, random_state=random_state)
    fold_size = len(triples) // k_folds
    folds = []
    for k in range(k_folds):
        start = k * fold_size
        end   = start + fold_size
        folds.append(triples[start:end])

    # All users are common
    common_users = list(range(m_t))

    mae_scores, rmse_scores = [], []

    # 4) k-fold CV on target
    for fold_idx in range(k_folds):
        test_data  = folds[fold_idx]
        train_data = []
        for j in range(k_folds):
            if j != fold_idx:
                train_data.extend(folds[j])

        R_train = np.zeros_like(R_t)
        R_test  = np.zeros_like(R_t)
        for (u, i, val) in train_data:
            R_train[u, i] = val
        for (u, i, val) in test_data:
            R_test[u, i] = val

        # Transfer train
        W_target, H_target = train_nmf_transfer(
            R_target=R_train,
            W_source=W_source,
            H_source=H_source,
            common_users=common_users,
            common_items=common_items,
            n_factors=n_factors,
            alpha=alpha,
            gamma=gamma,
            max_iter=max_iter,
            random_state=random_state + fold_idx
        )

        # Evaluate on test fold
        R_pred = predict_nmf(W_target, H_target)
        mae_t, rmse_t = evaluate(R_test, R_pred)
        mae_scores.append(mae_t)
        rmse_scores.append(rmse_t)
        print(f"Fold {fold_idx+1}: MAE={mae_t:.4f}, RMSE={rmse_t:.4f}")

    # 5) Summaries + 95% CI
    mae_mean, rmse_mean = np.mean(mae_scores), np.mean(rmse_scores)
    mae_std, rmse_std   = np.std(mae_scores, ddof=1), np.std(rmse_scores, ddof=1)
    z_value = 1.96
    n_f     = float(k_folds)
    mae_ci  = z_value * (mae_std / np.sqrt(n_f))
    rmse_ci = z_value * (rmse_std / np.sqrt(n_f))

    print("\n=== Final 5-Fold Results (Target Domain) ===")
    print(f"MAE:  {mae_mean:.4f} ± {mae_ci:.4f}  (95% CI)")
    print(f"RMSE: {rmse_mean:.4f} ± {rmse_ci:.4f}  (95% CI)")


# =============================================================================
# Example usage
# =============================================================================
if __name__ == "__main__":
    # Suppose you already have:
    #   source_rating_matrix, target_rating_matrix
    # that differ in #items => e.g. shape (m, n_s) vs. shape (m, n_t)
    # BOTH must be nonnegative and have the same #users = m.

    print("=== Source Matrix ===")
    print(source_rating_matrix)
    print("\n=== Target Matrix ===")
    print(target_rating_matrix)

    # Cross-validate with parameters matching your MF-with-bias code
    cross_validate_transfer_nmf(
        source_rating_matrix=source_rating_matrix,
        target_rating_matrix=target_rating_matrix,
        n_factors=10,   # latent dimension
        alpha=0.01,     # L2 reg
        gamma=0.1,      # transfer penalty
        max_iter=20,    # epochs
        k_folds=5,
        random_state=42
    )
```

```python
import numpy as np
from sklearn.utils import shuffle

# =============================================================================
# Reusing your NMF + transfer code (unchanged)
# =============================================================================

def train_nmf_regularized(R_source, n_factors=10, alpha=0.01, max_iter=20, tol=1e-4, random_state=42):
    """
    Standard Lee & Seung multiplicative updates with L2 regularization.
    Returns (W_source, H_source).
    """
    if hasattr(R_source, "to_numpy"):
        R_source = R_source.to_numpy()

    np.random.seed(random_state)
    n_users, n_items = R_source.shape

    W_source = np.random.rand(n_users, n_factors)
    H_source = np.random.rand(n_factors, n_items)

    def frob_norm_sq(X):
        return np.sum(X**2)

    prev_loss = None
    for _ in range(max_iter):
        # Update H
        numerator_H   = W_source.T @ R_source
        denominator_H = (W_source.T @ W_source) @ H_source + alpha * H_source
        denominator_H[denominator_H == 0] = 1e-10
        H_source *= (numerator_H / denominator_H)

        # Update W
        numerator_W   = R_source @ H_source.T
        denominator_W = W_source @ (H_source @ H_source.T) + alpha * W_source
        denominator_W[denominator_W == 0] = 1e-10
        W_source *= (numerator_W / denominator_W)

        # Optional converge check
        WH = W_source @ H_source
        recon = 0.5 * frob_norm_sq(R_source - WH)
        reg   = 0.5 * alpha * (frob_norm_sq(W_source) + frob_norm_sq(H_source))
        loss  = recon + reg
        if prev_loss is not None and abs(prev_loss - loss) < tol:
            break
        prev_loss = loss

    return W_source, H_source


def train_nmf_transfer(
    R_target, W_source, H_source,
    common_users, common_items,
    n_factors=10,
    alpha=0.01,
    gamma=0.1,
    max_iter=20,
    tol=1e-4,
    random_state=123
):
    """
    Lee & Seung updates plus a penalty pulling W_target,H_target toward W_source,H_source
    on the rows/cols in common_users, common_items.
    """
    if hasattr(R_target, "to_numpy"):
        R_target = R_target.to_numpy()

    np.random.seed(random_state)
    m, n = R_target.shape

    W_target = np.random.rand(m, n_factors)
    H_target = np.random.rand(n_factors, n)

    user_mask = np.zeros(m)
    user_mask[common_users] = 1.0

    item_mask = np.zeros(n)
    item_mask[common_items] = 1.0

    def frob_norm_sq(X):
        return np.sum(X**2)

    prev_loss = None
    for _ in range(max_iter):
        # Update H
        numerator_H   = W_target.T @ R_target
        denominator_H = (W_target.T @ W_target) @ H_target + alpha * H_target

        # Transfer for common items
        numerator_H += gamma * (H_source * item_mask[np.newaxis, :])
        denominator_H += gamma * (H_target * item_mask[np.newaxis, :])

        denominator_H[denominator_H == 0] = 1e-10
        H_target *= (numerator_H / denominator_H)

        # Update W
        numerator_W   = R_target @ H_target.T
        denominator_W = W_target @ (H_target @ H_target.T) + alpha * W_target

        # Transfer for common users
        numerator_W += gamma * (W_source * user_mask[:, np.newaxis])
        denominator_W += gamma * (W_target * user_mask[:, np.newaxis])

        denominator_W[denominator_W == 0] = 1e-10
        W_target *= (numerator_W / denominator_W)

        # Optional: check for convergence
        WtHt = W_target @ H_target
        recon = 0.5 * frob_norm_sq(R_target - WtHt)
        reg   = 0.5 * alpha * (frob_norm_sq(W_target) + frob_norm_sq(H_target))
        W_diff = (W_target - W_source)[common_users]
        H_diff = (H_target - H_source)[:, common_items]
        trans  = 0.5 * gamma * (frob_norm_sq(W_diff) + frob_norm_sq(H_diff))
        loss   = recon + reg + trans
        if prev_loss is not None and abs(prev_loss - loss) < tol:
            break
        prev_loss = loss

    return W_target, H_target


def predict_nmf(W, H):
    """Compute R_pred = W @ H."""
    return W @ H

def evaluate(R_true, R_pred):
    """
    Evaluate only where R_true > 0, returning (MAE, RMSE).
    """
    mask = (R_true > 0)
    errors = R_pred[mask] - R_true[mask]
    mae = np.mean(np.abs(errors))
    rmse = np.sqrt(np.mean(errors**2))
    return mae, rmse

# =============================================================================
# CROSS-VALIDATION WITH PARTIAL FRACTIONS (20%, 40%, 60%, 80%)
# =============================================================================

def cross_validate_transfer_nmf(
    source_rating_matrix,
    target_rating_matrix,
    n_factors=10,
    alpha=0.01,
    gamma=0.1,
    max_iter=20,
    k_folds=5,
    random_state=42
):
    """
    1) Train (W_source,H_source) on the FULL source domain.
    2) Possibly slice columns to match item dimension if source & target differ.
    3) Perform k-fold cross-validation on the target domain:
         - For each fold:
             a) Build the FULL training set, do standard transfer.
             b) Also sub-sample partial fractions (20%, 40%, 60%, 80%) of the training set,
                train separate NMF models for each fraction, evaluate on the same test fold.
    4) Return performance stats for each fraction plus the full training set.
    """
    # 1) Train on Source
    print("=== Train NMF on Source ===")
    if hasattr(source_rating_matrix, "to_numpy"):
        R_s = source_rating_matrix.to_numpy()
    else:
        R_s = source_rating_matrix

    W_source, H_source = train_nmf_regularized(
        R_source=R_s,
        n_factors=n_factors,
        alpha=alpha,
        max_iter=max_iter,
        random_state=random_state
    )
    R_s_pred = predict_nmf(W_source, H_source)
    mae_s, rmse_s = evaluate(R_s, R_s_pred)
    print(f"Source Fit => MAE={mae_s:.4f}, RMSE={rmse_s:.4f}\n")

    # Convert target
    if hasattr(target_rating_matrix, "to_numpy"):
        R_t = target_rating_matrix.to_numpy()
    else:
        R_t = target_rating_matrix

    # 2) Handle item dimension mismatch
    m_s, n_s = R_s.shape
    m_t, n_t = R_t.shape
    assert m_s == m_t, "User mismatch between source & target!"

    if n_s > n_t:
        # Slice H_source so that columns match target
        H_source = H_source[:, :n_t]
        common_items = list(range(n_t))
    else:
        # Slice target to match n_s
        R_t = R_t[:, :n_s]
        common_items = list(range(n_s))

    m_t, n_t_sliced = R_t.shape
    common_users = list(range(m_t))

    # Build (u,i,r) for target
    target_triples = []
    for u in range(m_t):
        for i in range(n_t_sliced):
            if R_t[u, i] > 0:
                target_triples.append((u, i, R_t[u, i]))

    target_triples = shuffle(target_triples, random_state=random_state)
    fold_size = len(target_triples) // k_folds
    folds = []
    for k in range(k_folds):
        start = k * fold_size
        end   = start + fold_size
        folds.append(target_triples[start:end])

    # We'll store fold-level results for FULL and partial fractions
    full_mae,  full_rmse  = [], []
    frac_scores = {0.2: [], 0.4: [], 0.6: [], 0.8: []}

    # 3) K-Fold CV
    for fold_idx in range(k_folds):
        test_data  = folds[fold_idx]
        train_data = []
        for j in range(k_folds):
            if j != fold_idx:
                train_data.extend(folds[j])

        # Build FULL train/test matrices
        R_train_full = np.zeros_like(R_t)
        R_test = np.zeros_like(R_t)

        for (u, i, r) in train_data:
            R_train_full[u, i] = r
        for (u, i, r) in test_data:
            R_test[u, i] = r

        # --- A) Train on FULL set ---
        W_target_full, H_target_full = train_nmf_transfer(
            R_target=R_train_full,
            W_source=W_source, H_source=H_source,
            common_users=common_users, common_items=common_items,
            n_factors=n_factors, alpha=alpha, gamma=gamma,
            max_iter=max_iter,
            random_state=random_state+fold_idx
        )
        R_pred_full = predict_nmf(W_target_full, H_target_full)
        maeF, rmseF = evaluate(R_test, R_pred_full)
        full_mae.append(maeF)
        full_rmse.append(rmseF)

        print(f"Fold {fold_idx+1} (FULL):   MAE={maeF:.4f}, RMSE={rmseF:.4f}")

        # --- B) Partial fraction training (20%, 40%, 60%, 80%) ---
        rng = np.random.default_rng(random_state)
        indices = np.arange(len(train_data))
        rng.shuffle(indices)

        for frac in frac_scores.keys():
            cutoff = int(len(train_data) * frac)
            subset_indices = indices[:cutoff]
            partial_data = [train_data[i] for i in subset_indices]

            # Build partial train matrix
            R_train_part = np.zeros_like(R_t)
            for (u, i, r) in partial_data:
                R_train_part[u, i] = r

            W_target_part, H_target_part = train_nmf_transfer(
                R_target=R_train_part,
                W_source=W_source, H_source=H_source,
                common_users=common_users, common_items=common_items,
                n_factors=n_factors, alpha=alpha, gamma=gamma,
                max_iter=max_iter,
                random_state=random_state + 1000 + fold_idx
            )
            R_pred_part = predict_nmf(W_target_part, H_target_part)
            maeP, rmseP = evaluate(R_test, R_pred_part)
            frac_scores[frac].append((maeP, rmseP))
            print(f"Fold {fold_idx+1} ({int(frac*100)}%): MAE={maeP:.4f}, RMSE={rmseP:.4f}")

    # 4) Summaries
    full_mae_mean  = np.mean(full_mae)
    full_mae_std   = np.std(full_mae, ddof=1)
    full_rmse_mean = np.mean(full_rmse)
    full_rmse_std  = np.std(full_rmse, ddof=1)

    print("\n=== Final Results: FULL target training ===")
    print(f"MAE:  {full_mae_mean:.4f} ± {full_mae_std:.4f}")
    print(f"RMSE: {full_rmse_mean:.4f} ± {full_rmse_std:.4f}")

    for frac in sorted(frac_scores.keys()):
        mae_list = [x[0] for x in frac_scores[frac]]
        rmse_list = [x[1] for x in frac_scores[frac]]
        mae_mean  = np.mean(mae_list)
        mae_std   = np.std(mae_list, ddof=1)
        rmse_mean = np.mean(rmse_list)
        rmse_std  = np.std(rmse_list, ddof=1)
        print(f"\n=== Final Results: {int(frac*100)}% of target training ===")
        print(f"MAE:  {mae_mean:.4f} ± {mae_std:.4f}")
        print(f"RMSE: {rmse_mean:.4f} ± {rmse_std:.4f}")


# =============================================================================
# EXAMPLE USAGE
# =============================================================================
if __name__ == "__main__":
    # Suppose you have source_rating_matrix, target_rating_matrix
    # This call performs cross-validation and, within each fold,
    # trains on full data plus partial fractions (20%, 40%, 60%, 80%).
    cross_validate_transfer_nmf(
        source_rating_matrix=source_rating_matrix,
        target_rating_matrix=target_rating_matrix,
        n_factors=10,
        alpha=0.01,
        gamma=0.1,
        max_iter=20,
        k_folds=5,
        random_state=42
    )
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold

def mae_rmse(preds, truths):
    """
    Compute MAE and RMSE for given predictions and ground truths.
    Both preds and truths are 1D arrays of the same length (non-zero entries).
    """
    errors = preds - truths
    mae = np.mean(np.abs(errors))
    rmse = np.sqrt(np.mean(errors ** 2))
    return mae, rmse


class DMFModel:
    """
    A simplistic DMF-like model where:
      - We store user_factors and item_factors as the top-layer representations.
      - We combine them with an element-wise product and a final linear layer.

    This is a placeholder for the deeper architecture proposed by
    Xue et al. (2017). In practice, you'd replace this with a full
    multi-layer network, non-linearities, etc.
    """

    def __init__(self, n_users, n_items, n_factors=20, learning_rate=0.005, lambda_reg=0.01):
        self.n_users = n_users
        self.n_items = n_items
        self.n_factors = n_factors
        self.lr = learning_rate
        self.lambda_reg = lambda_reg

        # Initialize user/item factors (as if they were the final embedding layer).
        self.user_factors = 0.01 * np.random.randn(n_users, n_factors).astype(np.float32)
        self.item_factors = 0.01 * np.random.randn(n_items, n_factors).astype(np.float32)

        # Final layer weights for the element-wise product.
        self.final_w = 0.01 * np.random.randn(n_factors).astype(np.float32)
        self.final_b = 0.0

    def forward(self, u, i):
        """
        Given a user index u and item index i, compute the predicted rating.
        r̂(u, i) = (user_factors[u] * item_factors[i]) dot final_w + final_b
        """
        user_vec = self.user_factors[u]
        item_vec = self.item_factors[i]
        z = user_vec * item_vec  # element-wise product
        pred = z.dot(self.final_w) + self.final_b
        return pred

    def train_one_epoch(self, rating_matrix):
        """
        One epoch of training with (very) naive SGD on each nonzero rating.
        rating_matrix[u, i] > 0 => a known rating from user u to item i.
        """
        users, items = rating_matrix.nonzero()
        for u, i in zip(users, items):
            r_ui = rating_matrix[u, i]
            pred_ui = self.forward(u, i)

            # Error
            err = (pred_ui - r_ui)

            # Current copies of factor vectors for backprop
            user_vec = self.user_factors[u].copy()
            item_vec = self.item_factors[i].copy()

            # Gradients for final_w, final_b
            grad_final_w = err * (user_vec * item_vec)
            grad_final_b = err

            # Gradients for user_factors and item_factors
            grad_user_factors = err * (self.final_w * item_vec)
            grad_item_factors = err * (self.final_w * user_vec)

            # Add L2 regularization
            grad_user_factors += self.lambda_reg * user_vec
            grad_item_factors += self.lambda_reg * item_vec
            grad_final_w       += self.lambda_reg * self.final_w
            # (We skip regularizing final_b for simplicity.)

            # Update
            self.user_factors[u] -= self.lr * grad_user_factors
            self.item_factors[i] -= self.lr * grad_item_factors
            self.final_w         -= self.lr * grad_final_w
            self.final_b         -= self.lr * grad_final_b

    def predict_all(self, rating_matrix):
        """
        Predict all non-zero (user, item) entries in 'rating_matrix'.
        Return two arrays: preds (predicted) and truths (actual ratings).
        """
        users, items = rating_matrix.nonzero()
        preds = []
        truths = []
        for u, i in zip(users, items):
            preds.append(self.forward(u, i))
            truths.append(rating_matrix[u, i])
        return np.array(preds), np.array(truths)


def cross_domain_dmf(
    source_rating_matrix,
    target_rating_matrix,
    n_factors=20,
    lambda_reg=0.01,
    epochs_source=50,
    epochs_target=30,
    k_folds=5,
    learning_rate=0.005
):
    """
    1) Trains DMF on source_rating_matrix (full set).
    2) Transfers some learned parameters to a new DMF model for the target domain.
    3) Performs k-fold cross validation on the target domain to evaluate performance.

    Returns:
        A dict:
          {
             'mae':  (mean_mae,  std_mae),
             'rmse': (mean_rmse, std_rmse)
          }
    """

    # If the inputs are DataFrames, convert them to NumPy arrays
    if isinstance(source_rating_matrix, pd.DataFrame):
        source_rating_matrix = source_rating_matrix.values
    if isinstance(target_rating_matrix, pd.DataFrame):
        target_rating_matrix = target_rating_matrix.values

    # 1) Train a DMF model on the SOURCE domain
    num_users_source, num_items_source = source_rating_matrix.shape
    dmf_source = DMFModel(
        n_users=num_users_source,
        n_items=num_items_source,
        n_factors=n_factors,
        learning_rate=learning_rate,
        lambda_reg=lambda_reg
    )
    for _ in range(epochs_source):
        dmf_source.train_one_epoch(source_rating_matrix)

    # 2) Create a fresh DMF model for the TARGET domain and transfer some parameters
    num_users_target, num_items_target = target_rating_matrix.shape
    dmf_target = DMFModel(
        n_users=num_users_target,
        n_items=num_items_target,
        n_factors=n_factors,
        learning_rate=learning_rate,
        lambda_reg=lambda_reg
    )
    # Example: Transfer the final layer parameters
    dmf_target.final_w = dmf_source.final_w.copy()
    dmf_target.final_b = dmf_source.final_b

    # 3) K-Fold Cross Validation on the target domain
    kf = KFold(n_splits=k_folds, shuffle=True, random_state=42)

    mae_scores = []
    rmse_scores = []

    # Get all pairs of (user, item) that have a non-zero rating in the target domain
    non_zero_pairs = np.array(list(zip(*target_rating_matrix.nonzero())))

    for train_index, test_index in kf.split(non_zero_pairs):
        train_pairs = non_zero_pairs[train_index]
        test_pairs  = non_zero_pairs[test_index]

        # Build train_matrix and test_matrix from the splits
        train_matrix = np.zeros_like(target_rating_matrix)
        for (u, i) in train_pairs:
            train_matrix[u, i] = target_rating_matrix[u, i]

        test_matrix = np.zeros_like(target_rating_matrix)
        for (u, i) in test_pairs:
            test_matrix[u, i] = target_rating_matrix[u, i]

        # Re-initialize a fold-specific DMF model from the trained source model
        fold_model = DMFModel(
            n_users=num_users_target,
            n_items=num_items_target,
            n_factors=n_factors,
            learning_rate=learning_rate,
            lambda_reg=lambda_reg
        )
        fold_model.final_w = dmf_source.final_w.copy()
        fold_model.final_b = dmf_source.final_b

        # Train on the current fold's train split
        for _ in range(epochs_target):
            fold_model.train_one_epoch(train_matrix)

        # Test on the current fold's test split
        preds, truths = fold_model.predict_all(test_matrix)
        fold_mae, fold_rmse = mae_rmse(preds, truths)

        mae_scores.append(fold_mae)
        rmse_scores.append(fold_rmse)

    results = {
        'mae':  (np.mean(mae_scores), np.std(mae_scores)),
        'rmse': (np.mean(rmse_scores), np.std(rmse_scores))
    }
    return results


# USAGE EXAMPLE (commented out because in real usage you'd pass your own data)
# if __name__ == "__main__":
#     # 1) Load or define your source_rating_matrix and target_rating_matrix.
#     #    shape = (num_users, num_items), 0 = missing, > 0 = observed rating
#     #
#     # Example:
#     # source_rating_matrix = pd.DataFrame(...)  # or a NumPy array
#     # target_rating_matrix = pd.DataFrame(...)  # or a NumPy array
#
#     # 2) Pass them directly to the function:
results = cross_domain_dmf(
         source_rating_matrix,  # could be a DataFrame or NumPy array
        target_rating_matrix,  # could be a DataFrame or NumPy array
         n_factors=20,
       lambda_reg=0.01,
        epochs_source=20,
         epochs_target=20,
         k_folds=5,
         learning_rate=0.005
 )
#
print("\n=== Xue et al. DMF Results ===")
print(f"MAE:  {results['mae'][0]:.4f} ± {results['mae'][1]:.4f}")
print(f"RMSE: {results['rmse'][0]:.4f} ± {results['rmse'][1]:.4f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold

def mae_rmse(preds, truths):
    """
    Compute MAE and RMSE for given predictions and ground truths.
    Both preds and truths are 1D arrays of the same length (non-zero entries).
    """
    errors = preds - truths
    mae = np.mean(np.abs(errors))
    rmse = np.sqrt(np.mean(errors ** 2))
    return mae, rmse


class DMFModel:
    """
    A simplistic DMF-like model where:
      - We store user_factors and item_factors as the top-layer representations.
      - We combine them with an element-wise product and a final linear layer.
    """

    def __init__(self, n_users, n_items, n_factors=20, learning_rate=0.005, lambda_reg=0.01):
        self.n_users = n_users
        self.n_items = n_items
        self.n_factors = n_factors
        self.lr = learning_rate
        self.lambda_reg = lambda_reg

        # Initialize user/item factors (like final embedding layer).
        self.user_factors = 0.01 * np.random.randn(n_users, n_factors).astype(np.float32)
        self.item_factors = 0.01 * np.random.randn(n_items, n_factors).astype(np.float32)

        # Final layer weights for the element-wise product.
        self.final_w = 0.01 * np.random.randn(n_factors).astype(np.float32)
        self.final_b = 0.0

    def forward(self, u, i):
        """
        Given a user index u and item index i, compute the predicted rating:
          r̂(u, i) = (user_factors[u] * item_factors[i]) dot final_w + final_b
        """
        user_vec = self.user_factors[u]
        item_vec = self.item_factors[i]
        z = user_vec * item_vec  # element-wise product
        pred = z.dot(self.final_w) + self.final_b
        return pred

    def train_one_epoch(self, rating_matrix):
        """
        One epoch of training with naive SGD on each nonzero rating.
        rating_matrix[u, i] > 0 => known rating from user u to item i.
        """
        users, items = rating_matrix.nonzero()
        for u, i in zip(users, items):
            r_ui = rating_matrix[u, i]
            pred_ui = self.forward(u, i)

            # Error
            err = (pred_ui - r_ui)

            # Copies for backprop
            user_vec = self.user_factors[u].copy()
            item_vec = self.item_factors[i].copy()

            # Gradients for final_w, final_b
            grad_final_w = err * (user_vec * item_vec)
            grad_final_b = err

            # Gradients for user_factors and item_factors
            grad_user_factors = err * (self.final_w * item_vec)
            grad_item_factors = err * (self.final_w * user_vec)

            # L2 regularization
            grad_user_factors += self.lambda_reg * user_vec
            grad_item_factors += self.lambda_reg * item_vec
            grad_final_w       += self.lambda_reg * self.final_w
            # (Skipping reg on final_b for simplicity.)

            # Updates
            self.user_factors[u] -= self.lr * grad_user_factors
            self.item_factors[i] -= self.lr * grad_item_factors
            self.final_w         -= self.lr * grad_final_w
            self.final_b         -= self.lr * grad_final_b

    def predict_all(self, rating_matrix):
        """
        Predict all non-zero (u, i) entries in 'rating_matrix'.
        Return two arrays: preds, truths
        """
        users, items = rating_matrix.nonzero()
        preds = []
        truths = []
        for u, i in zip(users, items):
            preds.append(self.forward(u, i))
            truths.append(rating_matrix[u, i])
        return np.array(preds), np.array(truths)


def cross_domain_dmf(
    source_rating_matrix,
    target_rating_matrix,
    n_factors=20,
    lambda_reg=0.01,
    epochs_source=50,
    epochs_target=30,
    k_folds=5,
    learning_rate=0.005
):
    """
    1) Trains DMF on source_rating_matrix (full set).
    2) Transfers final layer parameters to a new DMF model for the target domain.
    3) Performs k-fold cross validation on the target domain to evaluate performance
       on (A) 100% of the training set, and (B) partial fractions (20%, 40%, 60%, 80%).

    Returns a dictionary with mean/stdev MAE, RMSE for full-data training,
    plus fraction-based results.
    """

    # If the inputs are DataFrames, convert them to NumPy arrays
    if isinstance(source_rating_matrix, pd.DataFrame):
        source_rating_matrix = source_rating_matrix.values
    if isinstance(target_rating_matrix, pd.DataFrame):
        target_rating_matrix = target_rating_matrix.values

    # ----------------
    # 1) TRAIN on SOURCE
    # ----------------
    num_users_source, num_items_source = source_rating_matrix.shape
    dmf_source = DMFModel(
        n_users=num_users_source,
        n_items=num_items_source,
        n_factors=n_factors,
        learning_rate=learning_rate,
        lambda_reg=lambda_reg
    )
    print("Training DMF on SOURCE domain...")
    for _ in range(epochs_source):
        dmf_source.train_one_epoch(source_rating_matrix)
    print("Source-domain training complete.\n")

    # ----------------
    # 2) K-Fold on TARGET
    # ----------------
    num_users_target, num_items_target = target_rating_matrix.shape

    # We'll gather results for the FULL training set
    mae_scores_full = []
    rmse_scores_full = []

    # We'll gather results for partial fractions
    fractions = [0.2, 0.4, 0.6, 0.8]
    fraction_mae = {f: [] for f in fractions}
    fraction_rmse = {f: [] for f in fractions}

    # Nonzero user-item pairs in the target domain
    non_zero_pairs = np.array(list(zip(*target_rating_matrix.nonzero())))

    kf = KFold(n_splits=k_folds, shuffle=True, random_state=42)
    fold_id = 1

    for train_index, test_index in kf.split(non_zero_pairs):
        print(f"=== Fold {fold_id}/{k_folds} ===")
        fold_id += 1

        train_pairs = non_zero_pairs[train_index]
        test_pairs  = non_zero_pairs[test_index]

        # Build train_matrix and test_matrix
        train_matrix_full = np.zeros_like(target_rating_matrix)
        for (u, i) in train_pairs:
            train_matrix_full[u, i] = target_rating_matrix[u, i]

        test_matrix = np.zeros_like(target_rating_matrix)
        for (u, i) in test_pairs:
            test_matrix[u, i] = target_rating_matrix[u, i]

        # --- A) Full training on this fold's train set ---
        fold_model_full = DMFModel(
            n_users=num_users_target,
            n_items=num_items_target,
            n_factors=n_factors,
            learning_rate=learning_rate,
            lambda_reg=lambda_reg
        )
        # Transfer final layer from source
        fold_model_full.final_w = dmf_source.final_w.copy()
        fold_model_full.final_b = dmf_source.final_b

        # Train on FULL train_matrix
        for _ in range(epochs_target):
            fold_model_full.train_one_epoch(train_matrix_full)

        # Evaluate on test
        preds_full, truths_full = fold_model_full.predict_all(test_matrix)
        mae_full, rmse_full = mae_rmse(preds_full, truths_full)
        mae_scores_full.append(mae_full)
        rmse_scores_full.append(rmse_full)
        print(f" -> FULL data:    MAE={mae_full:.4f}, RMSE={rmse_full:.4f}")

        # --- B) Partial-fraction training ---
        # We'll shuffle train_pairs so the sub-sampling is random
        rng = np.random.default_rng(42)
        indices = np.arange(len(train_pairs))
        rng.shuffle(indices)  # in-place shuffle

        for frac in fractions:
            cutoff = int(len(train_pairs) * frac)
            partial_indices = indices[:cutoff]
            partial_pairs = train_pairs[partial_indices]

            # Build partial training matrix
            train_matrix_part = np.zeros_like(target_rating_matrix)
            for (u, i) in partial_pairs:
                train_matrix_part[u, i] = target_rating_matrix[u, i]

            # Fresh model for partial fraction
            fold_model_part = DMFModel(
                n_users=num_users_target,
                n_items=num_items_target,
                n_factors=n_factors,
                learning_rate=learning_rate,
                lambda_reg=lambda_reg
            )
            # Transfer final layer from source
            fold_model_part.final_w = dmf_source.final_w.copy()
            fold_model_part.final_b = dmf_source.final_b

            # Train on partial matrix
            for _ in range(epochs_target):
                fold_model_part.train_one_epoch(train_matrix_part)

            # Evaluate on the same test set
            preds_part, truths_part = fold_model_part.predict_all(test_matrix)
            mae_part, rmse_part = mae_rmse(preds_part, truths_part)

            fraction_mae[frac].append(mae_part)
            fraction_rmse[frac].append(rmse_part)
            print(f" -> {int(frac*100)}% data: MAE={mae_part:.4f}, RMSE={rmse_part:.4f}")

    # 4) Summaries
    def summarize_scores(scores):
        mean_val = np.mean(scores)
        std_val  = np.std(scores, ddof=1)
        return mean_val, std_val

    results = {}

    # Full data results
    mae_mean_full, mae_std_full = summarize_scores(mae_scores_full)
    rmse_mean_full, rmse_std_full = summarize_scores(rmse_scores_full)
    results['full'] = {
        'mae':  (mae_mean_full,  mae_std_full),
        'rmse': (rmse_mean_full, rmse_std_full)
    }

    # Fraction results
    for frac in fractions:
        mae_mean, mae_std = summarize_scores(fraction_mae[frac])
        rmse_mean, rmse_std = summarize_scores(fraction_rmse[frac])
        results[f'frac_{int(frac*100)}'] = {
            'mae':  (mae_mean,  mae_std),
            'rmse': (rmse_mean, rmse_std)
        }

    return results


# USAGE EXAMPLE (assuming you have source_rating_matrix and target_rating_matrix)
results = cross_domain_dmf(
    source_rating_matrix=source_rating_matrix,
    target_rating_matrix=target_rating_matrix,
    n_factors=20,
    lambda_reg=0.01,
    epochs_source=20,
    epochs_target=20,
    k_folds=5,
    learning_rate=0.005
)

print("\n=== DMF (Full) Results ===")
mae_mean_full, mae_std_full = results['full']['mae']
rmse_mean_full, rmse_std_full = results['full']['rmse']
print(f"MAE:  {mae_mean_full:.4f} ± {mae_std_full:.4f}")
print(f"RMSE: {rmse_mean_full:.4f} ± {rmse_std_full:.4f}")

for key in ['frac_20', 'frac_40', 'frac_60', 'frac_80']:
    mae_mean, mae_std  = results[key]['mae']
    rmse_mean, rmse_std = results[key]['rmse']
    print(f"\n=== DMF ({key} of training data) ===")
    print(f"MAE:  {mae_mean:.4f} ± {mae_std:.4f}")
    print(f"RMSE: {rmse_mean:.4f} ± {rmse_std:.4f}")
```

# SVD

```python
import numpy as np
from sklearn.model_selection import KFold

# हाइपरपैरामीटर्स
n_factors = 10
lambda_reg = 0.01
learning_rate = 0.005
epochs_target = 20
gamma = 0.1
k_folds = 5
train_fractions = [0.2, 0.4, 0.6, 0.8]  # नए ट्रेनिंग सेट आकार

class DirectTransferMF:
    def __init__(self, source_matrix, target_matrix):
        self.source = source_matrix.to_numpy() if isinstance(source_matrix, pd.DataFrame) else source_matrix
        self.target = target_matrix.to_numpy() if isinstance(target_matrix, pd.DataFrame) else target_matrix
        self.n_factors = n_factors
        self.reg = lambda_reg
        self.lr = learning_rate
        self.gamma = gamma

        # SVD initialization
        U, _, Vt = np.linalg.svd(self.source, full_matrices=False)
        self.U_source = U[:, :n_factors]
        self.V_source = Vt[:n_factors, :].T

    def _transfer_embeddings(self, source_emb, target_size):
        if source_emb.shape[0] >= target_size:
            return source_emb[:target_size] * self.gamma
        new_emb = np.zeros((target_size, n_factors))
        new_emb[:source_emb.shape[0]] = source_emb * self.gamma
        return new_emb

    def fit(self, train_indices):
        # Initialize target embeddings
        self.U_target = self._transfer_embeddings(self.U_source, self.target.shape[0])
        self.V_target = self._transfer_embeddings(self.V_source, self.target.shape[1])

        # SGD Training
        rows, cols = train_indices
        ratings = self.target[rows, cols]
        for epoch in range(epochs_target):
            for idx in np.random.permutation(len(ratings)):
                u, v, r = rows[idx], cols[idx], ratings[idx]
                error = r - self.U_target[u].dot(self.V_target[v])
                self.U_target[u] += self.lr * (error * self.V_target[v] - self.reg * self.U_target[u])
                self.V_target[v] += self.lr * (error * self.U_target[u] - self.reg * self.V_target[v])
        return self

    def predict(self, user_idx, item_idx):
        return np.clip(self.U_target[user_idx].dot(self.V_target[item_idx]), 1, 5)

def evaluate_transfer_learning(source_df, target_df):
    # Convert to numpy arrays
    source_matrix = source_df.to_numpy()
    target_matrix = target_df.to_numpy()

    # Get non-zero indices
    nnz_rows, nnz_cols = np.nonzero(target_matrix)
    ratings = target_matrix[nnz_rows, nnz_cols]

    # Results storage
    results = {frac: {'mae': [], 'rmse': []} for frac in train_fractions}

    # Cross-validation
    kf = KFold(n_splits=k_folds)
    for fold, (train_idx, test_idx) in enumerate(kf.split(ratings)):
        print(f"\nFold {fold+1}/{k_folds}")

        for frac in train_fractions:
            # Partial training data
            partial_size = int(len(train_idx) * frac)
            partial_idx = np.random.choice(train_idx, partial_size, replace=False)

            # Train model
            model = DirectTransferMF(source_df, target_df)
            model.fit((nnz_rows[partial_idx], nnz_cols[partial_idx]))

            # Predictions
            test_vals = target_matrix[nnz_rows[test_idx], nnz_cols[test_idx]]
            preds = [model.predict(u,i) for u,i in zip(nnz_rows[test_idx], nnz_cols[test_idx])]

            # Calculate metrics
            mae = np.mean(np.abs(test_vals - preds))
            rmse = np.sqrt(np.mean((test_vals - preds)**2))

            results[frac]['mae'].append(mae)
            results[frac]['rmse'].append(rmse)

            print(f"TR({int(frac*100)}%): MAE={mae:.3f}, RMSE={rmse:.3f}")

    # Final report
    print("\n=== Transfer Learning Effectiveness Analysis ===")
    for frac in train_fractions:
        mae_mean = np.mean(results[frac]['mae'])
        mae_ci = 1.96 * np.std(results[frac]['mae']) / np.sqrt(k_folds)
        rmse_mean = np.mean(results[frac]['rmse'])
        rmse_ci = 1.96 * np.std(results[frac]['rmse']) / np.sqrt(k_folds)

        print(f"\nTR({int(frac*100)}%) Results:")
        print(f"MAE  : {mae_mean:.4f} ± {mae_ci:.4f}")
        print(f"RMSE : {rmse_mean:.4f} ± {rmse_ci:.4f}")

# उपयोग उदाहरण
if __name__ == "__main__":
    evaluate_transfer_learning(source_rating_matrix, target_rating_matrix)
```

```python
import numpy as np
from sklearn.model_selection import KFold



source_rating_matrix = source_rating_matrix
target_rating_matrix = target_rating_matrix


def als_factorization(
    rating_matrix: np.ndarray,
    n_factors: int,
    lambda_reg: float,
    epochs: int,
    user_init: np.ndarray = None,
    item_init: np.ndarray = None
):
    num_users, num_items = rating_matrix.shape
    observed = ~np.isnan(rating_matrix)  # True wherever rating is known

    # Initialize
    if user_init is not None:
        user_factors = user_init.copy()
    else:
        user_factors = 0.01 * np.random.randn(num_users, n_factors)

    if item_init is not None:
        item_factors = item_init.copy()
    else:
        item_factors = 0.01 * np.random.randn(num_items, n_factors)

    for _ in range(epochs):
        # Update user factors
        for u in range(num_users):
            idx_items = np.where(observed[u])[0]
            if len(idx_items) == 0:
                continue
            item_factors_sub = item_factors[idx_items]
            ratings_sub = rating_matrix[u, idx_items]
            A = item_factors_sub.T @ item_factors_sub + lambda_reg * np.eye(n_factors)
            B = item_factors_sub.T @ ratings_sub
            user_factors[u, :] = np.linalg.solve(A, B)

        # Update item factors
        for i in range(num_items):
            idx_users = np.where(observed[:, i])[0]
            if len(idx_users) == 0:
                continue
            user_factors_sub = user_factors[idx_users]
            ratings_sub = rating_matrix[idx_users, i]
            A = user_factors_sub.T @ user_factors_sub + lambda_reg * np.eye(n_factors)
            B = user_factors_sub.T @ ratings_sub
            item_factors[i, :] = np.linalg.solve(A, B)

    return user_factors, item_factors


def cross_domain_recommender(
    source_rating_matrix: np.ndarray,
    target_rating_matrix: np.ndarray,
    n_factors: int = 10,
    lambda_reg: float = 0.01,
    gamma: float = 0.1,           # Not used in basic ALS, but left for extension
    learning_rate: float = 0.005, # Not used in basic ALS, but left for extension
    epochs_source: int = 20,
    epochs_target: int = 20,
    k_folds: int = 5
):
    # 1) Train on source domain
    user_factors_source, item_factors_source = als_factorization(
        source_rating_matrix, n_factors, lambda_reg, epochs_source
    )

    # 2) Transfer user factors for overlapping users (dummy approach)
    num_users_target, num_items_target = target_rating_matrix.shape
    user_factors_target_init = 0.01 * np.random.randn(num_users_target, n_factors)
    overlap_users = min(user_factors_source.shape[0], num_users_target)
    user_factors_target_init[:overlap_users, :] = user_factors_source[:overlap_users, :]
    item_factors_target_init = 0.01 * np.random.randn(num_items_target, n_factors)

    # 3) K-fold cross-validation in target domain
    user_item_pairs = np.argwhere(~np.isnan(target_rating_matrix))
    ratings = target_rating_matrix[~np.isnan(target_rating_matrix)]

    kf = KFold(n_splits=k_folds, shuffle=True, random_state=42)
    rmses, maes = [], []

    for train_index, test_index in kf.split(user_item_pairs):
        train_pairs = user_item_pairs[train_index]
        train_ratings = ratings[train_index]
        test_pairs = user_item_pairs[test_index]
        test_ratings = ratings[test_index]

        # Build a train matrix
        train_matrix = np.full_like(target_rating_matrix, np.nan, dtype=float)
        for (u, i), r in zip(train_pairs, train_ratings):
            train_matrix[u, i] = r

        # Fine-tune on the target domain
        user_factors_target, item_factors_target = als_factorization(
            train_matrix,
            n_factors,
            lambda_reg,
            epochs_target,
            user_init=user_factors_target_init.copy(),
            item_init=item_factors_target_init.copy()
        )

        # Evaluate
        preds = []
        for (u, i) in test_pairs:
            preds.append(user_factors_target[u] @ item_factors_target[i])
        preds = np.array(preds)
        fold_rmse = np.sqrt(np.mean((preds - test_ratings) ** 2))
        fold_mae = np.mean(np.abs(preds - test_ratings))

        rmses.append(fold_rmse)
        maes.append(fold_mae)

    # 4) Compute mean + 95% confidence interval
    def mean_confidence_interval(data, confidence=0.95):
        arr = np.array(data)
        mean_val = np.mean(arr)
        std_val = np.std(arr, ddof=1)
        n = len(arr)
        # normal approximation for 95% CI
        margin = 1.96 * (std_val / np.sqrt(n))
        return mean_val, margin

    rmse_mean, rmse_margin = mean_confidence_interval(rmses)
    mae_mean, mae_margin = mean_confidence_interval(maes)

    return {
        "RMSE_mean": rmse_mean,
        "RMSE_95%_CI": (rmse_mean - rmse_margin, rmse_mean + rmse_margin),
        "MAE_mean": mae_mean,
        "MAE_95%_CI": (mae_mean - mae_margin, mae_mean + mae_margin),
    }


# Convert DataFrames to NumPy arrays if they are not already
source_rating_matrix = source_rating_matrix.to_numpy()  # If they're DataFrames
target_rating_matrix = target_rating_matrix.to_numpy()

# Proceed with the rest of the code
results = cross_domain_recommender(
    source_rating_matrix,
    target_rating_matrix,
    n_factors=10,
    lambda_reg=0.01,
    epochs_source=20,
    epochs_target=20,
    k_folds=5
)

# Extract results from dictionary
mae_mean = results["MAE_mean"]
mae_lower_ci, mae_upper_ci = results["MAE_95%_CI"]
rmse_mean = results["RMSE_mean"]
rmse_lower_ci, rmse_upper_ci = results["RMSE_95%_CI"]

# Compute ± margin for printing
mae_margin  = (mae_upper_ci  - mae_mean)
rmse_margin = (rmse_upper_ci - rmse_mean)

print("\n=== Final 5-Fold Results (Target Domain) ===")
print(f"MAE:  {mae_mean:.4f} ± {mae_margin:.4f}  (95% CI)")
print(f"RMSE: {rmse_mean:.4f} ± {rmse_margin:.4f}  (95% CI)")
```

```python
import numpy as np
from sklearn.model_selection import KFold

def als_factorization(
    rating_matrix: np.ndarray,
    n_factors: int,
    lambda_reg: float,
    epochs: int,
    user_init: np.ndarray = None,
    item_init: np.ndarray = None
):
    """
    Performs ALS factorization for a given rating matrix.
    rating_matrix: shape (num_users, num_items), np.nan where no rating is present
    n_factors: number of latent factors
    lambda_reg: regularization parameter
    epochs: number of alternating iterations
    user_init, item_init: optional initial guesses for user/item factor matrices
    """
    num_users, num_items = rating_matrix.shape
    observed = ~np.isnan(rating_matrix)  # True wherever rating is known

    # Initialize user factors
    if user_init is not None:
        user_factors = user_init.copy()
    else:
        user_factors = 0.01 * np.random.randn(num_users, n_factors)

    # Initialize item factors
    if item_init is not None:
        item_factors = item_init.copy()
    else:
        item_factors = 0.01 * np.random.randn(num_items, n_factors)

    for _ in range(epochs):
        # Update user factors
        for u in range(num_users):
            idx_items = np.where(observed[u])[0]
            if len(idx_items) == 0:
                continue
            item_factors_sub = item_factors[idx_items]
            ratings_sub = rating_matrix[u, idx_items]
            A = item_factors_sub.T @ item_factors_sub + lambda_reg * np.eye(n_factors)
            B = item_factors_sub.T @ ratings_sub
            user_factors[u, :] = np.linalg.solve(A, B)

        # Update item factors
        for i in range(num_items):
            idx_users = np.where(observed[:, i])[0]
            if len(idx_users) == 0:
                continue
            user_factors_sub = user_factors[idx_users]
            ratings_sub = rating_matrix[idx_users, i]
            A = user_factors_sub.T @ user_factors_sub + lambda_reg * np.eye(n_factors)
            B = user_factors_sub.T @ ratings_sub
            item_factors[i, :] = np.linalg.solve(A, B)

    return user_factors, item_factors


def mean_confidence_interval(data, confidence=0.95):
    """
    Compute mean and 95% CI (normal approximation) for a list of scores.
    """
    arr = np.array(data)
    mean_val = np.mean(arr)
    std_val = np.std(arr, ddof=1)
    n = len(arr)
    margin = 1.96 * (std_val / np.sqrt(n))
    return mean_val, margin


def cross_domain_recommender(
    source_rating_matrix: np.ndarray,
    target_rating_matrix: np.ndarray,
    n_factors: int = 10,
    lambda_reg: float = 0.01,
    gamma: float = 0.1,           # Not used in basic ALS, left for extension
    learning_rate: float = 0.005, # Not used in basic ALS, left for extension
    epochs_source: int = 20,
    epochs_target: int = 20,
    k_folds: int = 5
):
    """
    1) Trains ALS on the source domain.
    2) Transfers overlapping user factors to initialize target domain factors.
    3) Performs K-fold CV on the target domain.
    4) Returns RMSE/MAE means and 95% confidence intervals.

    In addition, for validating transfer-learning effectiveness, we train
    on varying fractions of the training set within each fold: 20%, 40%, 60%, 80%.
    """
    # ----- Step 1: Train on source domain -----
    user_factors_source, item_factors_source = als_factorization(
        source_rating_matrix, n_factors, lambda_reg, epochs_source
    )

    # ----- Step 2: Transfer user factors for overlapping users (simple approach) -----
    num_users_target, num_items_target = target_rating_matrix.shape
    user_factors_target_init = 0.01 * np.random.randn(num_users_target, n_factors)
    overlap_users = min(user_factors_source.shape[0], num_users_target)
    user_factors_target_init[:overlap_users, :] = user_factors_source[:overlap_users, :]
    item_factors_target_init = 0.01 * np.random.randn(num_items_target, n_factors)

    # ----- Step 3: K-fold cross-validation in the target domain -----
    user_item_pairs = np.argwhere(~np.isnan(target_rating_matrix))
    ratings = target_rating_matrix[~np.isnan(target_rating_matrix)]

    kf = KFold(n_splits=k_folds, shuffle=True, random_state=42)

    # We'll track full-training-set performance
    full_rmse_scores, full_mae_scores = [], []

    # Also track performance for partial fractions of the training data
    fractions = [0.2, 0.4, 0.6, 0.8]
    fraction_results = {
        f: {"rmse": [], "mae": []}
        for f in fractions
    }

    for train_index, test_index in kf.split(user_item_pairs):
        # Train pairs/ratings
        train_pairs = user_item_pairs[train_index]
        train_ratings = ratings[train_index]

        # Test pairs/ratings
        test_pairs = user_item_pairs[test_index]
        test_ratings = ratings[test_index]

        # Build full training matrix
        train_matrix_full = np.full_like(target_rating_matrix, np.nan, dtype=float)
        for (u, i), r in zip(train_pairs, train_ratings):
            train_matrix_full[u, i] = r

        # --- 3A: Evaluate using the FULL training set for this fold ---
        user_factors_target_full, item_factors_target_full = als_factorization(
            train_matrix_full,
            n_factors,
            lambda_reg,
            epochs_target,
            user_init=user_factors_target_init.copy(),
            item_init=item_factors_target_init.copy()
        )

        preds_full = []
        for (u, i) in test_pairs:
            preds_full.append(user_factors_target_full[u] @ item_factors_target_full[i])
        preds_full = np.array(preds_full)

        fold_rmse_full = np.sqrt(np.mean((preds_full - test_ratings) ** 2))
        fold_mae_full = np.mean(np.abs(preds_full - test_ratings))

        full_rmse_scores.append(fold_rmse_full)
        full_mae_scores.append(fold_mae_full)

        # --- 3B: Evaluate using PARTIAL fractions of the training set ---
        # Shuffle train_pairs so that sub-sampling is random.
        rng = np.random.default_rng(42)  # or another seed if desired
        shuffled_indices = np.arange(len(train_pairs))
        rng.shuffle(shuffled_indices)

        for frac in fractions:
            cutoff = int(len(train_pairs) * frac)
            sub_indices = shuffled_indices[:cutoff]

            # Construct partial training matrix
            train_matrix_partial = np.full_like(target_rating_matrix, np.nan, dtype=float)
            for idx in sub_indices:
                (u, i) = train_pairs[idx]
                r = train_ratings[idx]
                train_matrix_partial[u, i] = r

            # Fine-tune on partial training matrix
            user_factors_partial, item_factors_partial = als_factorization(
                train_matrix_partial,
                n_factors,
                lambda_reg,
                epochs_target,
                user_init=user_factors_target_init.copy(),
                item_init=item_factors_target_init.copy()
            )

            # Evaluate partial model
            preds_partial = []
            for (u, i) in test_pairs:
                preds_partial.append(user_factors_partial[u] @ item_factors_partial[i])
            preds_partial = np.array(preds_partial)

            fold_rmse_partial = np.sqrt(np.mean((preds_partial - test_ratings) ** 2))
            fold_mae_partial = np.mean(np.abs(preds_partial - test_ratings))

            fraction_results[frac]["rmse"].append(fold_rmse_partial)
            fraction_results[frac]["mae"].append(fold_mae_partial)

    # ----- Step 4: Summaries -----
    # Full training results
    rmse_mean_full, rmse_margin_full = mean_confidence_interval(full_rmse_scores)
    mae_mean_full, mae_margin_full   = mean_confidence_interval(full_mae_scores)

    # Fraction results
    fraction_summaries = {}
    for frac in fractions:
        frac_rmse = fraction_results[frac]["rmse"]
        frac_mae  = fraction_results[frac]["mae"]

        rmse_mean_frac, rmse_margin_frac = mean_confidence_interval(frac_rmse)
        mae_mean_frac, mae_margin_frac   = mean_confidence_interval(frac_mae)

        fraction_summaries[frac] = {
            "rmse_mean": rmse_mean_frac,
            "rmse_95%_CI": (rmse_mean_frac - rmse_margin_frac, rmse_mean_frac + rmse_margin_frac),
            "mae_mean": mae_mean_frac,
            "mae_95%_CI": (mae_mean_frac - mae_margin_frac, mae_mean_frac + mae_margin_frac),
        }

    return {
        "full_training": {
            "RMSE_mean": rmse_mean_full,
            "RMSE_95%_CI": (rmse_mean_full - rmse_margin_full, rmse_mean_full + rmse_margin_full),
            "MAE_mean": mae_mean_full,
            "MAE_95%_CI": (mae_mean_full - mae_margin_full, mae_mean_full + mae_margin_full),
        },
        "fraction_summaries": fraction_summaries
    }

# ------------------------------
# Example usage (as in your code)
# ------------------------------

# Suppose 'source_rating_matrix' and 'target_rating_matrix' are Pandas DataFrames.
# Convert them to NumPy arrays:
source_rating_matrix = source_rating_matrix.to_numpy()
target_rating_matrix = target_rating_matrix.to_numpy()

results = cross_domain_recommender(
    source_rating_matrix,
    target_rating_matrix,
    n_factors=10,
    lambda_reg=0.01,
    epochs_source=20,
    epochs_target=20,
    k_folds=5
)

# Unpack the final results
full_results = results["full_training"]
print("\n=== FINAL 5-Fold Results with FULL Training (Target Domain) ===")
print(f"MAE:  {full_results['MAE_mean']:.4f} ± {full_results['MAE_95%_CI'][1] - full_results['MAE_mean']:.4f}  (95% CI)")
print(f"RMSE: {full_results['RMSE_mean']:.4f} ± {full_results['RMSE_95%_CI'][1] - full_results['RMSE_mean']:.4f}  (95% CI)")

print("\n=== RESULTS by Fraction of Target-Domain Training Data ===")
for frac, summary in results["fraction_summaries"].items():
    mae_mean = summary["mae_mean"]
    mae_lower, mae_upper = summary["mae_95%_CI"]
    rmse_mean = summary["rmse_mean"]
    rmse_lower, rmse_upper = summary["rmse_95%_CI"]

    print(f"\nTraining fraction: {int(frac*100)}%")
    print(f"MAE:  {mae_mean:.4f} ± {mae_upper - mae_mean:.4f}  (95% CI)")
    print(f"RMSE: {rmse_mean:.4f} ± {rmse_upper - rmse_mean:.4f}  (95% CI)")
```
